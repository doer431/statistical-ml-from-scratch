# Linkage Methods in Hierarchical Clustering

This notebook compares different linkage strategies
used to compute inter-cluster distances.


## What is Linkage?

Linkage defines how the distance between two clusters
is computed during the merging process.


```python
import pandas as pd
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import dendrogram, linkage
```

```python
data = pd.read_csv("datasets/03_H.csv")
X = data.values

```

```python
plt.scatter(X[:, 0], X[:, 1])
plt.title("Raw Data Distribution")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()
```

## Linkage Methods

- **Single:** Minimum distance between points
- **Complete:** Maximum distance between points
- **Average:** Mean distance
- **Ward:** Minimizes variance


```python
Z = linkage(X, method="single")
plt.figure(figsize=(7, 4))
dendrogram(Z)
plt.title("Single Linkage Dendrogram")
plt.show()
```

```python
Z = linkage(X, method="complete")
plt.figure(figsize=(7, 4))
dendrogram(Z)
plt.title("Complete Linkage Dendrogram")
plt.show()
```

```python
Z = linkage(X, method="average")
plt.figure(figsize=(7, 4))
dendrogram(Z)
plt.title("Complete Linkage Dendrogram")
plt.show()
```

```python
Z = linkage(X, method="ward")
plt.figure(figsize=(7, 4))
dendrogram(Z)
plt.title("Complete Linkage Dendrogram")
plt.show()
```

## Comparison Summary

| Linkage | Behavior | Suitable For |
|------|--------|-------------|
| Single | Chaining | Arbitrary shapes |
| Complete | Compact | Tight clusters |
| Average | Balanced | General use |
| Ward | Stable | Research & analytics |


## Conclusion

Ward linkage is preferred for producing compact,
well-separated clusters and interpretable dendrograms.
