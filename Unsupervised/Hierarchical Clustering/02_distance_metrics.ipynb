# Distance Metrics in Clustering

This notebook explores how different distance metrics quantify similarity
between data points and influence clustering behavior.


## Why Distance Metrics Matter

Clustering algorithms rely on distance calculations.
The choice of metric directly affects cluster shape and separation.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

## Distance Metrics

**Euclidean Distance**
\[
d(x,y) = \sqrt{\sum (x_i - y_i)^2}
\]

**Manhattan Distance**
\[
d(x,y) = \sum |x_i - y_i|
\]

**Minkowski Distance**
\[
d(x,y) = \left( \sum |x_i - y_i|^p \right)^{1/p}
\]


```python
def euclidean_distance(a, b):
    return np.sqrt(np.sum((a - b) ** 2))

def manhattan_distance(a, b):
    return np.sum(np.abs(a - b))

def minkowski_distance(a, b, p=2):
    return np.sum(np.abs(a - b) ** p) ** (1 / p)
```

```python
a = np.array([2, 3])
b = np.array([6, 7])

euclidean_distance(a, b), manhattan_distance(a, b), minkowski_distance(a, b, p=3)
```

### When to Use Which Distance

| Metric | Best Use Case |
|------|--------------|
| Euclidean | Compact numeric data |
| Manhattan | High-dimensional data |
| Minkowski | Flexible behavior |


## Observations

- Euclidean distance is sensitive to scale
- Manhattan distance reduces outlier influence
- Minkowski allows tuning via parameter p
