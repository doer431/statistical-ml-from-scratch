# Hierarchical Clustering from Scratch

**Category:** Unsupervised Learning  
**Algorithm:** Agglomerative Hierarchical Clustering  
**Distance Metric:** Euclidean  
**Linkage Method:** Ward  

This notebook demonstrates the complete hierarchical clustering pipeline
using a single, well-justified configuration.


## Objective

To understand how hierarchical clustering groups unlabeled data points
by progressively merging clusters and visualizing the process using a dendrogram.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import dendrogram, linkage
from sklearn.cluster import AgglomerativeClustering
```

## Dataset Description

The dataset consists of numeric features suitable for distance-based clustering.


## Load Data

```python
data = pd.read_csv("datasets/01_H.csv")
X = data.values
X.shape
```

## Raw Vizuvalization

```python
plt.scatter(X[:, 0], X[:, 1])
plt.title("Raw Data Distribution")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()
```

## Distance Metric: Euclidean

Euclidean distance measures straight-line distance between two points
in n-dimensional space and works well for compact, spherical clusters.


## Agglomerative Hierarchical Clustering Algorithm

1. Treat each data point as a separate cluster  
2. Compute distances between all clusters  
3. Merge the two closest clusters  
4. Update the distance matrix  
5. Repeat until a single cluster remains  

The merging process is visualized using a dendrogram.


```python
Z = linkage(X, method="ward")
plt.figure(figsize=(8, 5))
dendrogram(Z)
plt.title("Ward Linkage Dendrogram")
plt.xlabel("Data Points")
plt.ylabel("Distance")
plt.show()
```

## Choosing the Number of Clusters

The number of clusters is selected by identifying the largest vertical gap
in the dendrogram before cluster merges.


```python
model = AgglomerativeClustering(n_clusters=3, linkage="ward")
labels = model.fit_predict(X)

plt.scatter(X[:, 0], X[:, 1], c=labels)
plt.title("Final Clusters ")
plt.show()
```

## Conclusion

Hierarchical clustering provides an interpretable clustering structure
without requiring a predefined number of clusters.
Ward linkage produces compact and stable clusters.
